```python
def gcd(p, q):
    # Use Euclid's algorithm to find the GCD.
    while q != 0:
        p, q = q, p % q
    return p

# Define a function 'is_coprime' to check if two numbers are coprime (GCD is 1).
def is_coprime(x, y):
    # Check if the GCD of 'x' and 'y' is equal to 1.
    return gcd(x, y) == 1
```

```python
coef_range=10
RGB=[]
for p in range(-coef_range,coef_range):
    for q in range(1,coef_range):
        if is_coprime(p,q):
            for r in range(-coef_range,coef_range):
                for s in range(1,coef_range):
                    if is_coprime(r,s):
                        found=[]
                        M=snappy.Manifold('s776')
                        M.dehn_fill((p,q),0)
                        M.dehn_fill((r,s),1)
                        if M.identify()!=[]:
                            found.append((p,q,r,s))
                            found.append(M.identify()[0])
                            M.dehn_fill((0,0),1)
                        if found!=[]:
                            M.dehn_fill((r,s),2)
                            if M.identify()!=[]:
                                if M.identify()[0].name()!=found[1].name():
                                    found.append(M.identify()[0])
                                    print('Found an example:',found)
                                    RGB.append(found)
                            M.dehn_fill((0,0),2)
                        
```
