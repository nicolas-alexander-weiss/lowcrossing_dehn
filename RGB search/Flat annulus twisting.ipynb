```python
import snappy
import time
```

```python
L=snappy.Manifold('annulus1.lnk')
for m in range(-10,11):
    for n in range(-10,11):
        L.dehn_fill([(0,0),(1,n),(-1,n),(1,m)])
        K=L.filled_triangulation()
        if K.identify()!=[]:
            print(m,n,K.identify())
```
