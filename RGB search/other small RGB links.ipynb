```python
import snappy
import time
L=snappy.Manifold('RGB_half_twist.lnk')
```

```python
start_time=time.time()

same_surgery=[]
same_K14=[]

for a in range(-3,4):
    for b in range(-2,3):
        for c in range(-2,3):
            for d in range(-2,3):
                for e in range(-2,3):
                    for f in range(-2,3):
                        L.dehn_fill([(a+b,1),(0,0),(0,1),(1,-f),(1,c),(1,-b),(1,-d),(1,-e)])
                        K1=L.filled_triangulation()
                        L.dehn_fill([(a+b,1),(0,1),(0,0),(1,-f),(1,c),(1,-b),(1,-d),(1,-e)])
                        K2=L.filled_triangulation()
                        if K1.identify()!=[]:
                                if K2.identify()!=[]:
                                    if K1.identify()!=K2.identify():
                                        print(a,b,c,d,e,f,K1.identify(),K2.identify())
                                        same_surgery.append([a,b,c,d,e,f,K1.identify()[0].name(),K2.identify()[0].name()])
                                    if K1.identify()[0].name()==K2.identify()[0].name():
                                        if K1.identify()[0].name()=='K14n3155':
                                            print(a,b,c,d,e,f,K1.identify(),K2.identify())
                                            same_K14.append([a,b,c,d,e,f,K1.identify()[0].name(),K2.identify()[0].name()])
print('Minutes taken:',(time.time()-start_time)/60)
```
